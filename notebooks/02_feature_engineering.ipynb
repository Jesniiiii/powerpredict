```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/uci_household_5min.csv", index_col="datetime", parse_dates=True)

print(df.shape)
df.head()
```

```python
target_col = "Global_active_power"

for lag in [1, 2, 3, 6, 12]:
    df[f"{target_col}_lag_{lag}"] = df[target_col].shift(lag)

df.head(15)
```

```python
for window in [3, 6, 12]:
    df[f"{target_col}_roll_mean_{window}"] = df[target_col].shift(1).rolling(window).mean()
    df[f"{target_col}_roll_std_{window}"] = df[target_col].shift(1).rolling(window).std()

df.head(15)
```

```python
df["hour"] = df.index.hour
df["day_of_week"] = df.index.dayofweek
df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)

df[["hour", "day_of_week", "is_weekend", "hour_sin", "hour_cos"]].head(10)
```

```python
before_rows = len(df)
df = df.dropna()
after_rows = len(df)

print(f"Rows before: {before_rows}")
print(f"Rows dropped: {before_rows - after_rows}")
print(f"Rows remaining: {after_rows}")
```

```python
df["target_5min"] = df[target_col].shift(-1)
df["target_15min"] = df[target_col].shift(-3)
df["target_30min"] = df[target_col].shift(-6)

df = df.dropna()

df[[target_col, "target_5min", "target_15min", "target_30min"]].head(10)
```

```python
n = len(df)
train_end = int(n * 0.65)
val_end = int(n * 0.80)

train_df = df.iloc[:train_end]
val_df = df.iloc[train_end:val_end]
test_df = df.iloc[val_end:]

print(f"Train: {len(train_df)} rows ({train_df.index.min()} to {train_df.index.max()})")
print(f"Val:   {len(val_df)} rows ({val_df.index.min()} to {val_df.index.max()})")
print(f"Test:  {len(test_df)} rows ({test_df.index.min()} to {test_df.index.max()})")
```

```python
train_df.to_csv("../data/processed/train_uci_household.csv")
val_df.to_csv("../data/processed/val_uci_household.csv")
test_df.to_csv("../data/processed/test_uci_household.csv")

print("Saved train/val/test splits to data/processed/")
print(f"Feature columns: {list(df.columns)}")
```
